```python
import cv2

imagem = cv2.imread("mangueira.jpg")

if imagem is None:
    print("Erro ao carregar a imagem")
else:
    altura, largura = imagem.shape[:2]
    escalas = [2, 4, 8, 10]

    for escala in escalas:
        nova_largura = int(largura * escala)
        nova_altura = int(altura * escala)

        imagem_redimensionada = cv2.resize(
            imagem,
            (nova_largura, nova_altura),
            interpolation=cv2.INTER_LINEAR
        )

        nome_janela = f"Imagem {escala}x"

        cv2.namedWindow(nome_janela, cv2.WINDOW_AUTOSIZE)
        cv2.imshow(nome_janela, imagem_redimensionada)

        cv2.waitKey(0)  # mostra uma por vez
        cv2.destroyAllWindows()
```
